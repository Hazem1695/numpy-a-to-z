# **Phase 2 — Array Properties & Inspection**

**The Idea**

Every NumPy array has metadata (information about the array), not just values.

Think of it like this:

| Concept  | Meaning                         |
| -------- | ------------------------------- |
| Data     | The actual numbers              |
| Metadata | Shape, type, memory, dimensions |


# **1️⃣ Create One Base Array (we’ll use it for everything)**

````python
import numpy as np
````

````python
arr = np.array([[1,2,3],
               [4,5,6]], dtype=np.int32)
print(arr)
````

# **2️⃣ Core Properties (Understand Structure)**

1️⃣ `.shape` → Structure of array

````python
arr.shape
````

| Rows | Columns |
| ---- | ------- |
| 2    | 3       |


2️⃣ `.ndim` → Number of dimensions

````python
arr.ndim
````

| Array Type | ndim |
| ---------- | ---- |
| Scalar     | 0    |
| Vector     | 1    |
| Matrix     | 2    |
| Tensor     | 3+   |


3️⃣ `.size` → Total elements

````python
arr.size
````

Formula:

`size = rows × columns`

# **3️⃣ Data Type & Memory**
1️⃣ `.dtype` → Type of data

````python
arr.dtype
````

2️⃣ `.itemsize` → Size of ONE element (bytes)

````python
arr.itemsize
````

| Type    | Bytes |
| ------- | ----- |
| int32   | 4     |
| float64 | 8     |


3️⃣ `.nbytes` → Total memory used

````python
arr.nbytes
````

Formula:

`nbytes = size × itemsize`

# **4️⃣ Transform & View**

1️⃣ `.T` → Transpose (swap rows & columns)

````python
arr.T
````

| Before | After |
| ------ | ----- |
| (2,3)  | (3,2) |


2️⃣ `.flat` → Iterate over all elements

````python
for i in arr.flat:
  print(i)
````

Converts array → 1D iterator

# **5️⃣ Complex Numbers Support**

1️⃣ `.real` and `.imag`

````python
complex_arr = np.array([1+2j, 3+4j])

print(f"real part: {complex_arr.real}")
print(f"imaginary part: {complex_arr.imag}")
````

| Property | Meaning        |
| -------- | -------------- |
| real     | real part      |
| imag     | imaginary part |


# **6️⃣ Change Data Type**

1️⃣ `.astype()` → Convert type

````python
arr_float = arr.astype(float)
print(arr_float)
````

***⚠️ Important Note***

`arr.astype(float)`

❌ does NOT modify original array

✔️ returns a new one

# **FULL COMBINED EXAMPLE (Covers EVERYTHING)**

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]], dtype=np.int32)

# Core Properties
print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Size:", arr.size)

# Data Type & Memory
print("Data Type:", arr.dtype)
print("Item Size:", arr.itemsize)
print("Total Bytes:", arr.nbytes)

# Transform & View
print("\nTranspose:\n", arr.T)
print("\nFlat iteration:")
for x in arr.flat:
    print(x, end=" ")

# Complex Numbers Support
complex_arr = np.array([1+2j, 3+4j])
print("\n\nReal:", complex_arr.real)
print("Imag:", complex_arr.imag)

# Change Data Type
print("\nConverted to float:\n", arr.astype(float))
````

# **Comparison Table**

| Property | What it tells you | Example Output |
| -------- | ----------------- | -------------- |
| shape    | structure         | (2,3)          |
| ndim     | dimensions        | 2              |
| size     | total elements    | 6              |
| dtype    | data type         | int32          |
| itemsize | bytes per element | 4              |
| nbytes   | total memory      | 24             |
| T        | transpose         | (3,2)          |
| flat     | iterator          | 1,2,3,4,5,6    |
| real     | real part         | [1.,3.]        |
| imag     | imaginary part    | [2.,4.]        |
| astype() | convert type      | float array    |


# **How to Read ANY Array Instantly (YOUR GOAL)**

When you see any array, think like this:

1. Shape? → structure

2. ndim? → complexity

3. dtype? → type of data

4. size? → how big

5. memory? → itemsize + nbytes

6. need transform? → use `.T`

7. need iteration? → use `.flat`
