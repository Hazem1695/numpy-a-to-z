# **Phase 7 — Stacking & Splitting**

The Idea

You are combining and breaking arrays
| Action    | Meaning        |
| --------- | -------------- |
| Stacking  | combine arrays |
| Splitting | divide arrays  |


Base Arrays

````python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

A = np.array([[1,2,3],
              [4,5,6]])

B = np.array([[7,8,9],
              [10,11,12]])
````

````python
print(f"a: \n{a}\n")
print(f"b: \n{b}\n")
````

````python
print(f"A: \n{A}\n")
print(f"B: \n{B}\n")
````

# **1️⃣ `np.vstack()` (Vertical Stack)**

Definition: Stack arrays row-wise (on top of each other)

1️⃣ 1D Example

````python
print(np.vstack((a, b)))
````

2️⃣ 2D Example

````python
print(np.vstack((A, B)))
````

3️⃣ More Examples

````python
print(np.vstack((A, A)))
````

````python
print(np.vstack((B, A)))
````

**⚠️ Rule**

Arrays must have same number of columns

**Table**

| Input Shapes  | Result |
| ------------- | ------ |
| (3,) + (3,)   | (2,3)  |
| (2,3) + (2,3) | (4,3)  |


# **2️⃣ `np.hstack()` (Horizontal Stack)**

Definition: Stack arrays column-wise (side by side)

1️⃣ 1D Example

````python
print(np.hstack((a, b)))
````

2️⃣ 2D Example

````python
np.hstack((A, B))
````

3️⃣ More Examples

````python
print(np.hstack((A, A)))
````

````python
print(np.hstack((B, A)))
````

**⚠️ Rule**

Arrays must have same number of rows

**Table**
| Input Shapes  | Result |
| ------------- | ------ |
| (3,) + (3,)   | (6,)   |
| (2,3) + (2,3) | (2,6)  |


# **3️⃣ `np.concatenate()` (MOST FLEXIBLE)**

Definition: Join arrays along a specific axis

Syntax:

`np.concatenate((A, B), axis=0)`

EXAMPLES

1️⃣ Axis 0 (vertical)


````python
print(np.concatenate((A, B), axis=0))
````

same as `vstack`

2️⃣ Axis 1 (horizontal)

````python
print(np.concatenate((A, B), axis=1))
````

same as `hstack`

3️⃣ 1D Examples

````python
print(np.concatenate((a, b)))
````

4️⃣ More Advanced

````python
C = np.array([[100,200,300]])
print(C)
````

````python
print(np.concatenate((A, C), axis=0))
````

````python
print(np.concatenate((A, C), axis=1))
````

**⚠️ Rules**

| Axis   | Requirement  |
| ------ | ------------ |
| axis=0 | same columns |
| axis=1 | same rows    |
--------------------------------
**Table**
| Code   | Meaning    |
| ------ | ---------- |
| axis=0 | vertical   |
| axis=1 | horizontal |
--------------------------------
**vstack vs hstack vs concatenate**
| Function    | Direction  | Equivalent |
| ----------- | ---------- | ---------- |
| vstack      | vertical   | axis=0     |
| hstack      | horizontal | axis=1     |
| concatenate | flexible   | both       |


# **4️⃣ `np.split()` (Split Arrays)**

Definition: Divide array into parts

1️⃣ 1D Example



````python
arr = np.array([1,2,3,4,5,6])
print(np.split(arr, 3))
````

2️⃣ Split at specific indices

````python
arr1 = np.array([1,2,3,4,5,6,7,8])
np.split(arr1, [2,4])
````

3️⃣ 2D Examples

- 1️⃣ Split rows

````python
arr2 = np.arange(1,13).reshape(3,4)
print(arr2)
````

````python
print(np.split(arr, 3))
````

- 2️⃣ Split columns

````python
print(np.split(arr2, 2, axis=1))
````

4️⃣ More Examples

````python
print(np.split(arr2, [1,2]))
````

````python
print(np.split(arr2, [2], axis=1))
````

**⚠️ Rule**

Equal split required (unless using indices)

**Table**
| Code             | Meaning       |
| ---------------- | ------------- |
| split(arr,3)     | equal parts   |
| split(arr,[2,4]) | custom split  |
| axis=1           | split columns |


# **FULL COMBINED EXAMPLE**

````python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(f"a array: \n{a}\n")
print(f"b array: \n{b}\n")

A = np.array([[1,2,3],[4,5,6]])
B = np.array([[7,8,9],[10,11,12]])
print(f"A array: \n{A}\n")
print(f"B array: \n{B}\n")

# stacking
print(f"vstack: \n{np.vstack((A,B))}\n")
print(f"hstack: \n{np.hstack((A,B))}\n")

# concatenate
print(f"concatenate vertical: \n{np.concatenate((A,B), axis=0)}\n")
print(f"concatenate horizontal: \n{np.concatenate((A,B), axis=1)}\n")

# splitting
arr = np.arange(1,13)
print(f"splitting: \n{print(np.split(arr, 3))}\n")
````

# **COMPARISON TABLE**
| Concept            | Function       | Use          |
| ------------------ | -------------- | ------------ |
| Vertical combine   | vstack         | add rows     |
| Horizontal combine | hstack         | add columns  |
| Flexible combine   | concatenate    | full control |
| Split equal        | split          | equal parts  |
| Split custom       | split(indices) | custom       |
----------------------------------
# **HOW TO THINK**
| Goal           | Use            |
| -------------- | -------------- |
| add rows       | vstack         |
| add columns    | hstack         |
| full control   | concatenate    |
| divide equally | split          |
| divide custom  | split(indices) |


# **Practice**
```
A = np.arange(1,7).reshape(2,3)
B = np.arange(7,13).reshape(2,3)

# 1. Stack vertically
# 2. Stack horizontally
# 3. Concatenate axis=0
# 4. Concatenate axis=1
# 5. Split into 2 parts
# 6. Split columns into 3 parts
```

````python
A = np.arange(1,7).reshape(2,3)
print(f"A:\n{A}\n")
B = np.arange(7,13).reshape(2,3)
print(f"B:\n{B}\n")

# 1. Stack vertically
print(f"Stack vertically:\n{np.vstack((A, B))}\n")

# 2. Stack horizontally
print(f"Stack horizontally:\n{np.hstack((A, B))}\n")

# 3. Concatenate axis=0
print(f"Concatenate axis=0:\n{np.concatenate((A, B), axis=0)}\n")

# 4. Concatenate axis=1
print(f"Concatenate axis=1:\n{np.concatenate((A, B), axis=1)}\n")

# 5. Split into 2 parts
print(f"Split A into 2 parts:\n{np.split(A, [2], axis=1)}\n")
print(f"Split B into 2 parts:\n{np.split(B, 2, axis=0)}\n")

# 6. Split columns into 3 parts
print(f"Split A columns into 3 parts:\n{np.split(A, 3, axis=1)}")
````

# **FINAL CHEAT SHEET**
| Category      | Tools          |
| ------------- | -------------- |
| Stack rows    | vstack         |
| Stack columns | hstack         |
| Flexible join | concatenate    |
| Split equal   | split          |
| Split custom  | split(indices) |

# **1️⃣ `np.dstack()` — Depth Stack (3D)**

Definition: Stack arrays along 3rd axis (depth)

Converts 2D → 3D

Example

````python
A = np.array([[1,2,3],[4,5,6]])
B = np.array([[7,8,9],[10,11,12]])
print(A,"\n")
print(B)
````

````python
print(A.shape)
print(B.shape)
````

````python
dstack = np.dstack((A, B))
print(dstack)
````

````python
print(dstack.shape)
````

Visualization:

- Rows = 2
- Columns = 3
- Depth = 2 arrays

Think of it like:

- stacking layers like a sandwich 🥪
