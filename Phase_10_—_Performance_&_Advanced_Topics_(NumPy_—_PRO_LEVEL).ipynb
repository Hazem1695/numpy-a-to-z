# **1️⃣ Vectorization vs Loops**

Definition
| Concept       | Meaning                                                        |
| ------------- | -------------------------------------------------------------- |
| Vectorization | Applying operations on entire arrays at once (no Python loops) |
| Loops         | Iterating element by element using Python `for`                |
---
⚡ Why Vectorization is Faster?

- NumPy uses C under the hood
- Avoids Python overhead
- Uses SIMD & optimized memory access


1️⃣ Example 1: Basic Addition

❌ Loop

````python
import numpy as np
````

````python
arr = np.arange(1, 6)
print(arr,"\n")

result = []

for x in arr:
    result.append(int(x + 5))

print(result)
````

✅ Vectorized

````python
arr = np.arange(1, 6)
result = arr + 5
print(result)
````


2️⃣ Example 2: Multiply Arrays

❌ Loop

````python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a,"\n")
print(b,"\n")
result = []

for i in range(len(a)):
    result.append(int(a[i] * b[i]))

print(result)
````

✅ Vectorized

````python
result = a * b
print(result)
````

3️⃣ Example 3: Conditional Logic

❌ Loop

````python
arr = np.array([1,2,3,4,5])
print(arr,"\n")

result = []

for x in arr:
    if x > 3:
        result.append(int(x))

print(result)
````

✅ Vectorized

````python
result = arr[arr > 3]
print(result)
````

4️⃣ Example 4: Mathematical Functions

❌ Loop

````python
import math
arr = np.array([1,4,9])
print(arr,"\n")

result = []
for x in arr:
    result.append(math.sqrt(x))
print(result)
````

✅ Vectorized

````python
result = np.sqrt(arr)
print(result)
````

Comparison Table
| Feature         | Loops  | Vectorization |
| --------------- | ------ | ------------- |
| Speed           | Slow ❌ | Very Fast ✅   |
| Code Length     | Long ❌ | Short ✅       |
| Readability     | Medium | High          |
| Python Overhead | High ❌ | Low ✅         |


# **2️⃣ Memory Optimization**

Definition

Memory optimization = using **less RAM + efficient storage**


1️⃣ Example 1: Data Types


````python
arr = np.array([1,2,3], dtype=np.int64)
print(arr.dtype)
print(arr.nbytes)
````

Optimize:

````python
arr = np.array([1,2,3], dtype=np.int8)
print(arr.dtype)
print(arr.nbytes)
````

2️⃣ Example 2: Float Precision



````python
arr = np.array([1.0, 2.0], dtype=np.float64)
print(arr.dtype)
print(arr.nbytes)
````

Optimize:

````python
arr = arr.astype(np.float32)
print(arr.dtype)
print(arr.nbytes)
````

3️⃣ Example 3: Memory Size

````python
arr = np.arange(1000)
print(arr.dtype)
print(arr.nbytes)
````

4️⃣ Example 4: Avoid Copies

````python
a = np.arange(10)
b = a  # no new memory

# Verify they share the same memory
print(f"Are a and b the same object? {a is b}")
````

Data Type Comparison

| int32   | 4     | normal use     |
| int64   | 8     | large values   |

# **3️⃣ Copy vs View**

Definition
| Concept | Meaning                        |
| ------- | ------------------------------ |
| Copy    | New array (independent memory) |
| View    | Shares memory with original    |

1️⃣ Example 1: View


````python
a = np.array([1,2,3])
b = a.view()

b[0] = 100
print(f"a array: \n{a}\n")  # changed!
print(f"b array: \n{b}\n")
````

2️⃣ Example 2: Copy

````python
a = np.array([1,2,3])
b = a.copy()

b[0] = 100
print(f"a array: \n{a}\n")  # unchanged
print(f"b array: \n{b}\n")
````

3️⃣  Example 3: Slice = View


````python
a = np.arange(10)
b = a[0:5]

b[0] = 99

print(f"a array: \n{a}\n")
print(f"b array: \n{b}\n")
````

4️⃣ Example 4: Force Copy

````python
a = np.arange(10)
b = a[0:5].copy()

b[0] = 99

print(f"a array: \n{a}\n")
print(f"b array: \n{b}\n")
````

Comparison Table
| Feature         | Copy   | View   |
| --------------- | ------ | ------ |
| Memory          | New    | Shared |
| Speed           | Slower | Faster |
| Safe            | Yes    | No     |
| Reflect Changes | No     | Yes    |


# **4️⃣ Broadcasting**

Definition

Broadcasting = automatic expansion of arrays to match shapes

Rules:

1- Align shapes from right
2- Dimensions must be:
- Equal OR
- One of them is 1

1️⃣ Example 1: Scalar Broadcasting

````python
arr = np.array([1,2,3])
result = arr + 5
print(result)
````

2️⃣ Example 2: 1D + 2D

````python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20,30])

print(a + b)
````

3️⃣ Example 3: Column Broadcasting

````python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([[10],
              [20]])

print(a + b)
````

4️⃣ Example 4: Mismatch Error

````python
a = np.array([1,2,3])
b = np.array([1,2])

print(a + b)
# ERROR
````

5️⃣ Example 5: Fix Shape

````python
a = np.array([1,2,3])
b = np.array([1,2,1])

print(a + b)
````

Broadcasting Table
| Shape A | Shape B | Result |
| ------- | ------- | ------ |
| (3,)    | scalar  | (3,)   |
| (2,3)   | (3,)    | (2,3)  |
| (2,3)   | (2,1)   | (2,3)  |

---

Broadcasting Compatibility Table

| Array A Shape | Array B Shape | Resulting Shape | Status | Logic |
| :--- | :--- | :--- | :--- | :--- |
| `(3, 4)` | `(3, 4)` | `(3, 4)` | **Match** | Dimensions are identical. |
| `(3, 4)` | `(4,)` | `(3, 4)` | **Match** | `(4,)` is treated as `(1, 4)`. The `1` stretches to `3`. |
| `(3, 1)` | `(3, 4)` | `(3, 4)` | **Match** | The `1` in Array A stretches to match the `4`. |
| `(3, 1)` | `(1, 4)` | `(3, 4)` | **Match** | Both stretch: A becomes `(3, 4)`, B becomes `(3, 4)`. |
| `(3, 4)` | `(3,)` | **Fail** | **Error** | Trailing dimensions `4` and `3` don't match and neither is `1`. |




# **5️⃣ Structured Arrays (Optional)**

Definition

Array with named fields (like database rows)

1️⃣ Example 1: Basic

````python
# Define the structure
dtype = [('name', 'U10'), ('age', 'i4')]

# Create the array with data
arr = np.array([('Ali', 25), ('Omar', 30)], dtype=dtype)

print(arr['name'])
````

2️⃣ Example 2: Access Field

````python
print(arr['age'])
````

Example 3: Filter

````python
print(arr[arr['age'] > 26])
````

Example 4: Multiple Types

````python
# Define the structure
dtype = [('id', 'i4'), ('salary', 'f4')]

# Create the array with data
data = np.array([(101, 50000.50), (102, 62000.00), (103, 45500.75)], dtype=dtype)

print(f"Full Structured Array:\n{data}\n")

# Access by column name
print(f"All Salaries:\n{data['salary']}\n")

# Access a specific row
print(f"First Record: {data[0]}")
````

Comparison
| Feature    | Structured Array | Normal Array |
| ---------- | ---------------- | ------------ |
| Data Types | Mixed            | Same         |
| Fields     | Named            | No           |
| Use Case   | Tabular          | Numeric      |

# **NumPy Data Type Shorthand Codes**

| Code | Type | Description | Byte Size |
| :--- | :--- | :--- | :--- |
| `'b1'` | **Boolean** | True or False | 1 byte |
| `'i1'` | **Integer** | 8-bit signed integer | 1 byte |
| `'i2'` | **Integer** | 16-bit signed integer | 2 bytes |
| `'i4'` | **Integer** | 32-bit signed integer | 4 bytes |
| `'i8'` | **Integer** | 64-bit signed integer | 8 bytes |
| `'u4'` | **Unsigned Int** | 32-bit unsigned (positive only) | 4 bytes |
| `'f2'` | **Float** | Half-precision float | 2 bytes |
| `'f4'` | **Float** | Single-precision float | 4 bytes |
| `'f8'` | **Float** | Double-precision float | 8 bytes |
| `'c8'` | **Complex** | Complex number (two 32-bit floats) | 8 bytes |
| `'U10'` | **Unicode** | String of 10 Unicode characters | 40 bytes* |
| `'S10'` | **String** | Zero-terminated bytes (10 chars) | 10 bytes |
| `'M8'` | **Datetime** | 64-bit date/time | 8 bytes |

> **Note on Unicode:** In NumPy, `'U'` strings use 4 bytes per character to support the full Unicode range. So, `'U10'` actually reserves 40 bytes of memory per entry.

You can also specify how the bytes are ordered in memory by adding a prefix:
```
<i4: Little-endian (Standard for Intel/AMD processors)

>i4: Big-endian (Common in network protocols)
```
