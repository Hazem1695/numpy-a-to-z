2️⃣ np.random.randn()

- Normal (Gaussian) distribution
- Mean = 0, Std = 1

Example 1:

```python
print(np.random.randn())
```

Example 2:

```python
print(np.random.randn(5))
```

Example 3 (2D):

```python
print(np.random.randn(2,3))
```

Example 4 (scaled distribution):

```python
# Generates 1,000 random numbers following a Normal Distribution
# with a Mean (average) of 50 and a Standard Deviation of 10.
A = np.random.randn(1000) * 10 + 50
print(A)
```

$$X = \sigma Z + \mu$$

- $X$: The resulting General Normal Distribution (your variable A).
- $Z$: The Standard Normal Distribution (`the np.random.randn()` part).
- $\sigma$ (Sigma): The Standard Deviation (the 10).
- $\mu$ (Mu): The Mean (the 50).

---
Summary Table
| Component | Math Term | Effect on Data |
| :--- | :--- | :--- |
| **`randn(1000)`** | Distribution | Creates 1,000 points centered at $0$ |
| **`* 10`** | $\sigma$ (Sigma) | Stretches the curve (makes it wider) |
| **`+ 50`** | $\mu$ (Mu) | Moves the center of the curve to $50$ |
------

Why do this?

If you are building a machine learning model to predict house prices or student grades, you often need "fake" data that looks "real." Real-world data is almost always "Normal" (most values are near the average, with few extremes), and this formula is the quickest way to create it.

1️⃣ np.random.choice()

Important Rule ⚠️

Pick random values from array

Example 1:

```python
arr = [10, 20, 30, 40]
print(np.random.choice(arr))
```

Example 2 (multiple picks):

```python
print(np.random.choice(arr, size=3))
```

Example 3 (with probability):

```python
print(np.random.choice(arr, size=5, p=[0.1,0.2,0.3,0.4]))
```

`p=[0.1, 0.2, 0.3, 0.4]`: The specific probability for each item in arr.

- The 1st item has a 10% chance.

- The 2nd item has a 20% chance.

- The 3rd item has a 30% chance.

- The 4th item has a 40% chance (The "luckiest" item).

Example 4 (without replacement):

```python
print(np.random.choice(arr, size=3, replace=False))
```

`replace=False` (Unique Sampling)
This is like picking a marble out of a bag and keeping it in your pocket.

- Every item picked is unique.

- Once an item is chosen, it is removed from the "pool" for the next draw.

2️⃣ `np.random.binomial()`

Binary outcomes (success/failure)

Parameters:
- n = number of trials
- p = probability of success

binomial math formula:

$$\binom{n}{k} = \frac{n!}{k!(n-k)!}$$

formula for the probability of $k$ successes:

$$P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}$$

Summary Table

| Symbol | Name | Referent / Meaning |
| :--- | :--- | :--- |
| **$n$** | Trials | How many times did you try? |
| **$k$** | Successes | How many "wins" are you looking for? |
| **$p$** | Success Probability | How likely is a "win" in one try? |
| **$1-p$ ($q$)** | Failure Probability | How likely is a "loss" in one try? |
| **$\binom{n}{k}$** | Combinations | How many patterns of wins/losses exist? |

----

Example in Words:

If you are shooting **10 basketball free throws** ($n=10$) and your skill level is **70%** ($p=0.7$), what is the chance you make **exactly 8** ($k=8$)?

1.  $\binom{10}{8}$ counts the different sequences of makes/misses that result in 8 total makes.
2.  $0.7^8$ is the probability of the 8 makes.
3.  $0.3^2$ is the probability of the 2 misses.

-------
Example 1:

```python
print(np.random.binomial(10, 0.5))
```

Example 2:

```python
print(np.random.binomial(10, 0.2, size=5))
```

Example 3 (extreme probability):

```python
print(np.random.binomial(10, 0.9, size=5))
```

Example 4 (coin flip simulation):

```python
print(np.random.binomial(1, 0.5, size=10))
```

3️⃣ `np.random.poisson()`

counts events in fixed interval

The Mathematical Formula:

$$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$$

Summary Table

| Symbol | Name | Plain English Meaning |
| :--- | :--- | :--- |
| **$\lambda$** | Rate Parameter | "On average, how many usually happen?" |
| **$k$** | Occurrences | "How many am I betting will happen now?" |
| **$e^{-\lambda}$** | Decay Factor | Adjusts the probability based on the average rate. |
| **$k!$** | Factorial | Normalizes the probability as $k$ increases. |


Example 1:

```python
# 3: The average number of events (lambda) expected per interval.
print(np.random.poisson(3))
```

Example 2:

```python
print(np.random.poisson(5, size=10))
```

Example 3 (rare events):

```python
print(np.random.poisson(0.5, size=10))
```

Example 4 (large rate):

```python
print(np.random.poisson(20, size=10))
```

3️⃣ `np.std()` (Standard Deviation)

Measures spread

Standard Deviation represents the "average distance" of every data point from the center of the data. It is the square root of the Variance.

The Mathematical Formula:

$$\sigma = \sqrt{\frac{\sum_{i=1}^{n} (x_i - \mu)^2}{n}}$$


- $\sigma$ (Sigma): The Standard Deviation.
- $x_i$: Each individual value in your data.
- $\mu$ (Mu): The Mean (average) of all the data points.
- $n$: The total number of values in the dataset.
- $\sum$: The Greek symbol for "Sum" (add everything up).

Example 1:


```python
print(np.std([1,2,3,4]))
```

Example 2:

```python
A = np.array([[1,2],
              [3,4]])

np.std(A, axis=0)
```

Example 3 (compare spread):

```python
# Explanation: Every number is the same.
# There is NO "deviation" from the average (10).
# The data is 100% consistent.
print(np.std([10,10,10]),"\n")

# Explanation: The numbers are very far apart (1 to 100).
# The average is 37.0, but the points are nowhere near it.
# This indicates high "volatility" or a wide spread.
print(np.std([1,10,100]),"\n")
```

4️⃣ `np.var()` (Variance): Square of std

Variance measures how far a set of numbers is spread out from their average value.

- If Variance = 0, all values are identical.

- A small variance indicates that the data points tend to be very close to the mean.

- A high variance indicates that the data points are very spread out from the mean and from each other.

The Mathematical Formula

$$\sigma^2 = \frac{\sum_{i=1}^{n} (x_i - \mu)^2}{n}$$

- $\sigma^2$: The symbol for Variance (pronounced "Sigma squared").
- $\sum$: The sum of all elements.
- $x_i$: Each individual data point.
- $\mu$: The Mean (average) of the data.
- $n$: The total number of data points.



Example 1:


```python
print(np.var([1,2,3,4]))
```

Example 2:

```python
A = np.array([[1,2],
              [3,4]])

print(np.var(A, axis=1))
```

5️⃣ `np.percentile()`: Position in distribution

The $P$-th percentile is the value such that $P\%$ of the data is less than or equal to that value.
- 25th Percentile ($Q1$): 25% of data is below this point.
- 50th Percentile ($Q2$): 50% of data is below this point (This is the Median).
- 75th Percentile ($Q3$): 75% of data is below this point.

The Mathematical Formula

$$R = \frac{P}{100} \times (n - 1)$$

The Interpolation Formula:
$$\text{Percentile Value} = V_{low} + (V_{high} - V_{low}) \times \text{fraction}$$


Example 1:



```python
print(np.percentile([1,2,3,4,5], 50))
```

median


The 50th percentile is the "cut-off" point where 50% of the data lies below it and 50% lies above it. It represents the exact middle of your dataset.

Example 2:

```python
print(np.percentile([1,2,3,4,5], 25),"\n")
print(np.percentile([1,2,3,4,5], 75),"\n")
```

Example 3 (multiple percentiles):

```python
print(np.percentile([1,2,3,4,5], [25,50,75]))
```

Example 4 (axis):

```python
A = np.array([[1,2],
              [3,4]])

print(np.percentile(A, 50, axis=0))
```
