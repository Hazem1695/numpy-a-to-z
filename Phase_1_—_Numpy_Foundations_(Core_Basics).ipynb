# **1️⃣ What is NumPy & Why It’s FAST**

NumPy = Numerical Python → library for working with arrays and numerical computation.

------------------------------
**Why NumPy is FAST**

Key Idea: Continuous Memory

1- Python list → scattered in memory

`list = [1, 2, 3]`

Stored as:

`[ pointer → 1 ] [ pointer → 2 ] [ pointer → 3 ]`

2- NumPy array → stored in one continuous block

`
import numpy as np
arr = np.array([1, 2, 3])
`

Stored as:

`[ 1 ][ 2 ][ 3 ]`

Direct continuous block → faster access + better cache usage

**Comparison**

| Feature       | Python List | NumPy Array    |
| ------------- | ----------- | -------------- |
| Memory layout | Scattered   | Continuous     |
| Speed         | Slow        | Fast           |
| Data type     | Mixed       | Same type only |
| Operations    | Loop needed | Vectorized     |

-------------------------------

**Speed Comparison**


```python
import numpy as np
import time

# Python list
lst = list(range(1000000))
start = time.time()
lst = [x * 2 for x in lst]
print("List time:", time.time() - start)

# NumPy
arr = np.arange(1000000)
start = time.time()
arr = arr * 2
print("NumPy time:", time.time() - start)
```

# **3️⃣ Creating Arrays (ALL CASES)**

**1️⃣ np.array()**

```python
# 1D
a = np.array([1,2,3])
print(a)
```

```python
# 2D
a = np.array([[1,2],[3,4]])
print(a)
```

```python
# 3D
a = np.array([[[1,2], [3,4]], [[5,6], [7,8]]])
print(a)
```

Force dtype

```python
a = np.array([1,2,3], dtype=float)
print(a)
```

Copy vs Reference

```python
a = np.array([1, 2, 3])
b = np.array(a)   # copy

print(f"a before change {a}")

a[0] = 100

print(f"b array {b}")  # unchanged
print(f"a after change {a}")
```

# **4️⃣ Random Functions 🎲**

**1️⃣ Random values 🎲**

```python
# 1D
a = np.random.rand(3)
print(a)
```

```python
# 2D
a = np.random.rand(2,3)
print(a)
```

```python
# 3D
a = np.random.rand(2,3,3)
print(a)
```

**2️⃣ Random integers 🎲**

```python
a = np.random.randint(0,10) # give you random integers 0 to 10
print(a)
```

```python
# 1D
a = np.random.randint(0,10, size=5) # create 1D array of 5 random integers from 0 to 9
print(a)
```

```python
# 2D
a = np.random.randint(0, 10, size=(2,3)) # 2D array (2 rows × 3 columns) of random integers 0–9
print(a)
```

```python
# 3D
a = np.random.randint(0, 10, size=(2,3,3)) # 3D array (2 blocks × 3 rows × 3 columns) of random integers 0–9
print(a)
```

**3️⃣ Normal distribution 🎲**

```python
a = np.random.randn(3) # 1D array of 3 random numbers from standard normal distribution
print(a)
```

```python
a = np.random.randn(2,2) # 2D array (2×2) of random numbers from standard normal distribution
print(a)
```

```python
a = np.random.randn(3,2,2)  # 3D array (3 blocks × 2 rows × 2 columns) of random numbers from standard normal distribution
print(a)
```

**4️⃣ Choice 🎲**

```python
a = np.random.choice([1,2,3,4]) # Pick one random element from [1,2,3,4]
print(a)
```

```python
a = np.random.choice([1,2,3], size=5)  # Create 1D array of 5 random picks from [1,2,3]
print(a)
```

```python
a = np.random.choice([1,2,3], size=5, p=[0.1,0.3,0.6])  # Create 1D array of 5 random picks from [1,2,3] with given probabilities
print(a)
```

```python
a = np.random.choice([1,2,3], size=(5,3))  # Create 2D array (5 rows × 3 cols) of random picks from [1,2,3]
print(a)
```

```python
a = np.random.choice([1,2,3], size=(5,5), p=[0.1,0.3,0.6])  # Create 2D array (5×5) of random picks from [1,2,3] with given probabilities
print(a)
```

```python
a = np.random.choice([1,2,3], size=(2,3,3))  # Create 3D array (2 blocks × 3 rows × 3 cols) of random picks from [1,2,3]
print(a)
```

```python
a = np.random.choice([1,2,3], size=(2,3,3), p=[0.1,0.3,0.6])  # Create 3D array (2×3×3) of random picks with given probabilities
print(a)
```

**5️⃣ Shuffle 🎲**

```python
arr = np.array([1,2,3,4])
np.random.shuffle(arr)  # shuffle elements of arr in place
print(arr)
```

**6️⃣ Seed (IMPORTANT for reproducibility) 🎲**

- This ensures you get the same random numbers every time you run the code.


```python
np.random.seed(42) # set random seed for reproducibility
np.random.rand(3)
```

**Comparison**

| Function | Meaning                    |
| -------- | -------------------------- |
| rand     | uniform (0–1)              |
| randn    | normal distribution        |
| randint  | integers                   |
| choice   | random selection           |
| shuffle  | shuffle array (in-place)   |

# **5️⃣ Data Types (dtype)**

What is dtype?

Every NumPy array has one fixed type

**1️⃣ Check dtype**

```python
arr = np.array([1,2,3])
print(arr.dtype)
```

**2️⃣ Common types**

```python
np.array([1,2,3], dtype=np.int32)
np.array([0.1, 2.2, 6.3], dtype=np.float64)
np.array([1,0,1], dtype=bool)
np.array(['a','b'], dtype=str)
```

**3️⃣ Convert dtype**

```python
arr = np.array([1,2,3])
print(f"before convert: {arr.dtype}")
arr = arr.astype(float)
print(f"after convert: {arr.dtype}")
```

**4️⃣ Memory size**

```python
arr = np.array([1,2,3], dtype=np.int32)
print(f"bytes per element: {arr.itemsize}")  # bytes per element
print(f"total bytes: {arr.nbytes}")    # total bytes
```

**Common Types**
| Type              | Meaning    |
| ----------------- | ---------- |
| int32 / int64     | integers   |
| float32 / float64 | decimals   |
| bool              | True/False |
| object            | mixed      |
