# **Phase 4 — Array Operations**

The Idea

NumPy does operations:

- Element-wise (NOT like Python lists)
- Without loops (vectorization)
- With broadcasting (automatic expansion)

# **1️⃣ Arithmetic Operations (Element-wise)**

Base Arrays

````python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
````

1️⃣ Addition

````python
print(a + b)
````

👉 Element by element:

`1+10, 2+20, 3+30`

2️⃣ Subtraction

````python
print( b - a)
````

3️⃣ Multiplication

````python
print(a * b)
````

4️⃣ Division

````python
print(b / a)
````

5️⃣ Power

````python
print(a **2)
````

````python
print(b ** 2)
````

````python
print(b % a)
````

# **2️⃣ Element-wise vs Python Lists (VERY IMPORTANT)**

````python
# Python list
print([1,2,3] + [10,20,30])                      # [1,2,3,10,20,30] ❌
# NumPy
print(np.array([1,2,3]) + np.array([10,20,30]))  # [11 22 33] ✅
````

👉 NumPy = math

👉 Python list = concatenation

# **4️⃣ Broadcasting**

Definition

Broadcasting: NumPy automatically expands arrays to match shapes

--------------------------

### ***Rule (MUST MEMORIZE)***

Two arrays are compatible if:

1- Same shape ✅

2- One dimension is 1 ✅

-------------------------------
1️⃣ Example 1: Vector + Scalar


````python
a = np.array([1, 2, 3])

print(a + 10)
````

2️⃣ Example 2: 2D + Scalar

````python
arr = np.array([[1,2,3],
                [4,5,6]])
print( arr + 10)
````

3️⃣ Example 3: Row Broadcasting

````python
arr = np.array([[1,2,3],
                [4,5,6]])

row = np.array([10,20,30])

print(arr + row)
````

row applied to EACH row

4️⃣ Example 4: Column Broadcasting

````python
arr = np.array([[1,2,3],
                [4,5,6]])

col = np.array([[10],
                [20]])

print(arr + col)
````

column applied to EACH column

5️⃣ Example 5: Different Shapes

````python
a = np.array([[1,2,3]])
b = np.array([[1],[2],[3]])

print( a + b)
````

**Broadcasting Visualization**

| Shape A | Shape B | Result |
| ------- | ------- | ------ |
| (2,3)   | (3,)    | (2,3)  |
| (2,3)   | (2,1)   | (2,3)  |
| (1,3)   | (3,1)   | (3,3)  |


❌ Broadcasting Error

````python
a = np.array([1,2,3])
b = np.array([1,2])

print(a + b)
````

❌ Error (shapes incompatible)

# **FULL COMBINED EXAMPLE**

````python
arr = np.arange(1,7).reshape(2,3)
print(f"{arr}\n")

row = np.array([10,20,30])
print(f"{row}\n")

col = np.array([[1],[2]])
print(f"{col}\n")

print(f"row broadcast: {arr + row}\n")   # row broadcast

print(f"column broadcast: {arr + col}\n")   # column broadcast

print(f"scalar: {arr * 2}\n")     # scalar

print(f"square root: {np.sqrt(arr)}\n")
````

# **COMPARISON TABLE**

| Concept      | Description              | Example    |
| ------------ | ------------------------ | ---------- |
| Arithmetic   | basic math               | a + b      |
| Scalar       | one value applied to all | a + 5      |
| Broadcasting | shape expansion          | arr + row  |
| Element-wise | operation per element    | a * b      |
| Functions    | math ops                 | np.sqrt(a) |

# **HOW TO THINK**

When solving:

| Situation        | Solution         |
| ---------------- | ---------------- |
| same shape       | direct operation |
| scalar           | auto broadcast   |
| row vs matrix    | broadcast row    |
| column vs matrix | broadcast column |


# **Practice**

```
arr = np.arange(1,13).reshape(3,4)

# 1. Add 5 to all elements
# 2. Multiply each row by [1,2,3,4]
# 3. Add column vector [[1],[2],[3]]
# 4. Take sqrt of all elements
# 5. Subtract row [10,20,30,40]
```

````python
arr = np.arange(1,13).reshape(3,4)
print(f"{arr}\n")

# 1. Add 5 to all elements
print(f"{arr + 5}\n")

# 2. Multiply each row by [1,2,3,4]
arr1 = np.array([1, 2, 3, 4])
print(f"{arr * arr1}\n")

# 3. Add column vector [[1],[2],[3]]
arr2 = np.array([[1],[2],[3]])
print(f"{arr + arr2}\n")

# 4. Take sqrt of all elements
print(f"{np.sqrt(arr)}\n")

# 5. Subtract row [10,20,30,40]
arr3 = np.array([10, 20, 30, 40])
print(arr - arr3)
````

# **Final Advice (VERY IMPORTANT)**

👉 If you ever write a loop like this:

````python
arr = np.arange(1, 13).reshape(3, 4)
print(f"before using loop:\n {arr}\n")
for i in range(len(arr)):
    arr[i] = arr[i] + 5

print(f"after using loop:\n {arr}")
````

**❌ STOP immediately**

✔ Replace with:

````python
arr = np.arange(1, 13).reshape(3, 4)
print(f"{arr}\n")

print(arr + 5)
````
