# **Phase 5 — Mathematical Function**

The Idea

NumPy lets you:

- Aggregate data (sum, mean, etc.)
- Analyze distribution (std, var)
- Apply math functions fast
- Work without loops

Base Arrays

````python
import numpy as np

arr1 = np.array([1, 2, 3, 4, 5])

arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])
````

# **1️⃣ Aggregation Functions**

1️⃣ `np.sum()`

````python
print(np.sum(arr1))
print(np.sum(arr2))
print(np.sum(arr2, axis=0))
print(np.sum(arr2, axis=1))
````

Understanding `axis`

| Axis   | Meaning     |
| ------ | ----------- |
| axis=0 | column-wise |
| axis=1 | row-wise    |


2️⃣ `np.mean()`

````python
print(np.mean(arr1))
print(np.mean(arr2))
print(np.mean(arr2, axis=0))
print(np.mean(arr2, axis=1))
````

3️⃣ `np.median()`

````python
print(np.median(arr1))
print(np.median(arr2))
````

4️⃣ `np.std()` (Standard Deviation)

````python
print(np.std(arr1))
print(np.std(arr2))
````

measures spread of data

5️⃣ `np.var()` (Variance)

````python
print(np.var(arr1))
print(np.var(arr2))
````

variance = std²

**Aggregation Table**

| Function | Meaning        |
| -------- | -------------- |
| sum      | total          |
| mean     | average        |
| median   | middle value   |
| std      | spread         |
| var      | spread squared |



# **2️⃣ Min & Max**

````python
print(np.min(arr1))
print(np.max(arr1))

print(np.min(arr2))
print(np.max(arr2))

print(np.min(arr2, axis=0))
print(np.max(arr2, axis=0))

print(np.min(arr2, axis=1))
print(np.max(arr2, axis=1))
````

**Table**
| Function | Result         |
| -------- | -------------- |
| min      | smallest value |
| max      | largest value  |

# **4️⃣ Trigonometric Functions**



````python
arr = np.array([0, np.pi/2, np.pi])
print(arr)
````

````python
print(np.sin(arr))
print(np.cos(arr))
print(np.tan(arr))
````

⚠️**Important Note**: NumPy uses radians NOT degrees

Convert degrees → radians

````python
print(np.deg2rad(90))
````

Table

# **5️⃣ Rounding Functions**

````python
arr = np.array([1.2, 2.5, 3.7])
print(arr)
````

1️⃣ `np.round()`

````python
print(np.round(arr))
````

2️⃣ `np.floor()`

````python
print(np.floor(arr))
````

3️⃣ `np.ceil()`

````python
print(np.ceil(arr))
````

**Table**

| Function | Result  |
| -------- | ------- |
| round    | nearest |
| floor    | down    |
| ceil     | up      |


# **FULL COMBINED EXAMPLE**

````python
arr = np.arange(1,10).reshape(3,3)

print(f"Sum: {np.sum(arr)}\n")
print(f"Mean: {np.mean(arr)}\n")
print(f"Std: {np.std(arr)}\n")

print(f"Min: {np.min(arr)}\n")
print(f"Max: {np.max(arr)}\n")

print(f"Sqrt:\n {np.sqrt(arr)}\n")
print(f"Log:\n {np.log(arr)}\n")

print(f"Round:\n {np.round(np.sqrt(arr))}\n")
````

# **COMPARISON**
| Category    | Functions          |
| ----------- | ------------------ |
| Aggregation | sum, mean, median  |
| Spread      | std, var           |
| Min/Max     | min, max           |
| Math        | sqrt, exp, log     |
| Trig        | sin, cos, tan      |
| Rounding    | round, floor, ceil |


# **🧠 HOW TO THINK**
| Goal           | Use         |
| -------------- | ----------- |
| total          | sum         |
| average        | mean        |
| spread         | std         |
| middle value   | median      |
| transform data | sqrt/log    |
| clean numbers  | round/floor |


# **Practice**

```
arr = np.array([[10, 20, 30],
                [40, 50, 60]])

# 1. Sum each column
# 2. Mean of each row
# 3. Find max in each column
# 4. Apply sqrt to all values
# 5. Round the result
```

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])

# 1. Sum each column
print(f"Sum: {np.sum(arr, axis=0)}\n")

# 2. Mean of each row
print(f"Mean{np.mean(arr, axis=1)}\n")

# 3. Find max in each column
print(f"Max: {np.max(arr, axis=0)}\n")

# 4. Apply sqrt to all values
print(f"Sqrt: {np.sqrt(arr)}\n")

# 5. Round the result
print(f"Round:\n {np.round(np.sqrt(arr))}\n")
````
