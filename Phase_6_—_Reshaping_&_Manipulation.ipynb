# **Phase 6 — Reshaping & Manipulation**

The Idea

You are not changing the data, You are changing how the data is viewed/structured

Base Arrays

````python
import numpy as np

arr = np.arange(1, 13)
print(arr)
````

# **1️⃣ `reshape()`**

Idea

Change shape WITHOUT changing data

1️⃣ Basic Examples

````python
print(arr.reshape(3,4))
````

````python
print(arr.reshape(4,3))
````

2️⃣ 3D Example

````python
print(arr.reshape(2,2,3))
````

3️⃣ Auto dimension (`-1`)

````python
print(arr.reshape(3, -1))
````

shape = (3,4)

````python
print(arr.reshape(-1, 2))
````

shape = (6,2)

NumPy calculates automatically

4️⃣ Invalid reshape ❌

````python
print(arr.reshape(5,3))
````

ERROR → total elements must match

**reshape RULE TABLE**

| Rule                  | Meaning      |
| --------------------- | ------------ |
| Total elements same   | MUST         |
| Order preserved       | YES          |
| -1 allowed            | Auto compute |
| Can go to higher dims | YES          |


# **2️⃣ 2. flatten() — Convert to 1D (COPY)**

Concept

- Always returns a copy
- Original array NOT affected

1️⃣ EXAMPLES

````python
a = np.array([[1,2], [3,4]])
print(f"before flatten: \n{a}\n")

b = a.flatten()

print(f"after flatten: {b}")
````

````python
arr2 = arr.reshape(3,4)
print(f"before flatten: \n{arr2}\n")

flat = arr2.flatten()
print(f"before flatten: \n{flat}\n")
````

2️⃣ Modify result

````python
b[0] = 100
print(f"a array before modify: \n {a}\n")
print(f"b array before modify: {b}")
````

⚠️ DOES NOT change original array

# **3️⃣ ravel() — Convert to 1D (VIEW)**

Concept

- Returns view if possible
- Changes affect original array

1️⃣ EXAMPLES

````python
a = np.array([[1, 2], [3, 4]])
print(f"before ravel: \n{a}\n")

b = a.ravel()
print(f"before ravel: \n{b}\n")
````

````python
arr2 = arr.reshape(3,4)
print(f"before ravel: \n{arr2}\n")

ravel = arr2.ravel()
print(f"before ravel: \n{ravel}\n")
````

2️⃣ Modify result

````python
b[0] = 100
print(f"a array before modify: \n {a}\n")
print(f"b array before modify: \n {b}")
````

⚠️ CHANGES original array❗

Table

| Function | Memory | Effect            |
| -------- | ------ | ----------------- |
| flatten  | copy   | safe              |
| ravel    | view   | modifies original |


# **4️⃣ transpose() / `.T`**

Concept

- Convert rows → columns

1️⃣ 2D Example



````python
arr2 = arr.reshape(3,4)
print(f"arr2 before transpose:\n{arr2}\n")

print(f"arr2 after transpose:\n{arr2.T}")
````

Shape becomes (4,3)

More Examples

````python
print(np.transpose(arr2))
````

Same result

2️⃣ 3D Transpose

````python
x = np.arange(8).reshape(2,2,2)
print(f"x before transpose:\n{x}\n")

print(f"x after transpose:\n{np.transpose(x, (1,0,2))}\n")
````

swaps axes

````python
a = np.arange(8).reshape(2,2,2)
print(f"a before transpose:\n{a}\n")

b = a.transpose(1,0,2)
print(f"b after transpose:\n{b}\n")
````

Same result

**transpose RULE**

| Input Shape | Output Shape    |
| ----------- | --------------- |
| (2, 3)      | (3, 2)          |
| (a, b, c)   | depends on axes |


**Table**
| Function    | Meaning           |
| ----------- | ----------------- |
| .T          | swap rows/columns |
| transpose() | full control      |


# **5️⃣ Resize (VERY IMPORTANT ⚠️)**
Two types:

1️⃣ `np.resize()` (function)

Can repeat data

````python
a = np.array([1,2,3])
print(f"b before resize:\n{a}\n")

b = np.resize(a, (2,4))
print(f"b after resize:\n{b}\n")
````

2️⃣ `a.resize()` (method)

Expanding


````python
arr = np.array([1,2,3,4])
print(f"arr before resize:\n{arr}\n")

arr.resize(3,3)
print(f"arr after resize:\n{arr}\n")

arr.resize(3,3,3)
print(f"arr after second resize:\n{arr}\n")
````

**⚠️ fills with 0**

**⚠️ Important**

- modifies original array
- dangerous if not careful

**resize vs reshape**

| Feature      | reshape | resize |
| ------------ | ------- | ------ |
| Safe         | ✅       | ❌      |
| Changes data | ❌       | ✅      |
| Repeats data | ❌       | ✅      |


# **5️⃣ Axis Control**

Concept

Axis = direction

````python
a = np.array([[1,2,3],
              [4,5,6]])

    # Axis 0 (columns ↓)
print(np.sum(a, axis=0))
    # Axis 1 (rows →)
print(np.sum(a, axis=1))
````

**Axis Table**
| Axis | Meaning        |
| ---- | -------------- |
| 0    | Down (columns) |
| 1    | Across (rows)  |


1️⃣ Add dimension

````python
arr = np.arange(1,7).reshape(2,3)
print(arr)
print(arr.shape)
````

````python
a = arr[np.newaxis, :]
print(a.shape)
````

shape: (1,2,3)

````python
a = arr[:, np.newaxis]
print(a.shape)
````

shape: (2,1,3)

2️⃣ expand_dims()

````python
a = np.expand_dims(arr, axis=0) # add new axis at row
print(a.shape)
````

````python
a = np.expand_dims(arr, axis=1) # add new axis at column
print(a.shape)
````

3️⃣ Remove dimension

````python
x = np.array([[[1,2,3]]])
print(x.shape)
a = np.squeeze(x)
print(a.shape)
````

removes extra dimensions

**Table**

| Function    | Purpose    |
| ----------- | ---------- |
| newaxis     | add dim    |
| expand_dims | add dim    |
| squeeze     | remove dim |


# **6️⃣ Shape Control (Pro Level)**

1️⃣ `reshape` for ML

````python
X = np.array([1,2,3,4])
print(X, "\n")
print(X.reshape(-1,1),"\n")   # column vector
print(X.reshape(1,-1), "\n")   # row vector
````

**Why important?**

| Shape | Meaning |
| ----- | ------- |
| (n,)  | 1D      |
| (n,1) | column  |
| (1,n) | row     |


# **FULL COMBINED EXAMPLE**

````python
arr = np.arange(1,13)

# reshape
x = arr.reshape(3,4)

# flatten vs ravel
flat = x.flatten()
rav = x.ravel()

# transpose
t = x.T

# expand dims
new = x[np.newaxis, :]

# squeeze
sq = np.squeeze(new)

print(f"reshape: \n{x}\n")
print(f"flat: \n{flat}\n")
print(f"ravel: \n{rav}\n")
print(f"transpose: \n{t}\n")
print(f"expand dims: \n{new.shape}\n")
print(f"squeeze: \n{sq.shape}\n")
````

# **Practice**
```
arr = np.arange(1,25)

# 1. reshape to (4,6)
# 2. reshape to (2,3,4)
# 3. flatten it
# 4. use ravel and modify first element
# 5. transpose a (3,4) array
# 6. convert [1,2,3,4] to column vector
```



````python
arr = np.arange(1,25)

# 1. reshape to (4,6)
a = arr.reshape(4,6)
print(f"reshape arr(a):\n {a}\n")

# 2. reshape to (2,3,4)
b = arr.reshape(2,3,4)
print(f"reshape arr(b):\n {b}\n")

# 3. flatten it
flat_a = a.flatten()
flat_b = b.flatten()
print(f"flat_a: \n{flat_a}\n")
print(f"flat_b: \n{flat_b}\n")

# 4. use ravel and modify first element
ravel_a = a.ravel()
ravel_a[0] = 99
print(f"ravel_a: \n{ravel_a}\n")
ravel_b = b.ravel()
ravel_b[0] = 999
print(f"ravel_b: \n{ravel_b}\n")

# 5. transpose a (3,4) array
arr1 = np.arange(1,13).reshape(3,4)
print(f"transpose: \n{arr1.T}\n")

# 6. convert [1,2,3,4] to column vector
X = np.array([1,2,3,4])
print(f"X array before convert: \n{X}\n")
a = X.reshape(-1,1)
print(f"column vector: \n{a}\n")
````
